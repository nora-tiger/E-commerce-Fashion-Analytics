# Phase 2 — Data Cleaning and Pre-processing

After understanding the structure of the original dataset, the next step is to assess its quality and prepare it for analysis. The initial inspection identified missing values in several columns and showed that `Purchase Date` was stored in an unsuitable datatype.

In this phase, I check and handle missing values, examine duplicate records, correct datatypes, create useful derived features, perform basic filtering and aggregation, and finally validate the processed dataset.

The cleaning decisions are made only after examining the identified issues, rather than automatically deleting records or replacing values.

## 2.1 Missing-Value Assessment

The initial dataset inspection showed that some columns had fewer non-null values than the total number of records. I therefore calculated both the number and percentage of missing values in each column.

The missing-value count shows the extent of incomplete data, while the percentage makes it easier to understand the size of the issue relative to the 10,000 transaction records. These values are examined before deciding how the missing data should be handled.

```python
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

print("========== MISSING VALUES ==========")
print(missing_values)

print("\n========== MISSING VALUE PERCENTAGE ==========")
print(missing_percentage.round(2))
```

### Observation

Missing values are present in four columns:

- `Online Store` — 2,249 missing values (**22.49%**)
- `Size` — 1,035 missing values (**10.35%**)
- `Festival/Sale` — 6,788 missing values (**67.88%**)
- `Delivery Speed` — 2,249 missing values (**22.49%**)

All other original columns contain complete values at this stage.

The amount of missing data differs considerably across these variables, particularly for `Festival/Sale`, where more than half of the records are missing. Removing every row containing a missing value would therefore discard a substantial amount of transaction data.

### Decision

The affected columns are categorical rather than continuous numerical measures. Instead of immediately deleting the corresponding transactions, the missing entries will be retained as explicit categories so that the records remain available for analysis.

## 2.2 Missing-Value Treatment

Since the missing values occur in categorical fields, I used descriptive labels rather than numerical imputation methods such as mean or median.

For `Online Store` and `Delivery Speed`, the missing entries are represented as `Not Applicable`. For `Size` and `Festival/Sale`, the missing entries are represented as `Not Specified`.

This approach preserves all 10,000 transaction records and clearly separates originally unavailable information from the existing categories. In particular, `Not Specified` is used for `Festival/Sale` rather than assuming that every missing value definitely represents a non-sale transaction.

```python
df["Online Store"] = df["Online Store"].fillna("Not Applicable")
df["Delivery Speed"] = df["Delivery Speed"].fillna("Not Applicable")
df["Size"] = df["Size"].fillna("Not Specified")
df["Festival/Sale"] = df["Festival/Sale"].fillna("Not Specified")
```

### Validation of Missing-Value Treatment

After applying the treatment, I checked the affected columns again to confirm that the missing values had been replaced successfully.

```python
print("========== MISSING VALUES AFTER TREATMENT ==========")
print(
    df[
        ["Online Store", "Size", "Festival/Sale", "Delivery Speed"]
    ].isnull().sum()
)
```

### Result

The validation confirms that no actual null values remain in the four treated columns.

The missing-value treatment therefore retained the transaction records while making previously unavailable categorical information explicit. A complete dataset-wide missing-value validation is also performed at the end of Phase 2.

## 2.3 Duplicate Record Check

Duplicate records can cause the same transaction to be counted more than once and may distort summary statistics or grouped results. I therefore checked whether any complete row was an exact duplicate of an earlier record.

Duplicate rows are checked before removing anything, since repeated purchases or multiple transactions by the same customer are not automatically duplicates.

```python
duplicate_count = df.duplicated().sum()

print("========== DUPLICATE RECORDS ==========")
print("Number of Exact Duplicate Rows:", duplicate_count)
```

### Observation and Decision

The dataset contains **0 exact duplicate rows**.

Therefore, no duplicate records need to be removed. Multiple transactions or repeated purchases are retained because they can represent genuine customer activity rather than duplicated data.

## 2.4 Datatype Correction

The initial datatype check showed that `Purchase Date` was stored as an object rather than as a datetime variable. Keeping a date as plain text would limit time-based analysis and make it difficult to extract information such as purchase year and month.

I therefore converted `Purchase Date` to a proper datetime datatype. Invalid or unrecognized date values, if present, are converted to `NaT` so that they can be detected rather than causing the conversion to fail.

```python
df["Purchase Date"] = pd.to_datetime(
    df["Purchase Date"],
    errors="coerce"
)

print("Purchase Date datatype:", df["Purchase Date"].dtype)
print("Invalid / missing dates after conversion:", df["Purchase Date"].isna().sum())
```

### Observation

`Purchase Date` was successfully converted to `datetime64[ns]`, and the validation showed that no missing or invalid dates remained after conversion.

### Decision

The corrected datetime format is retained because it supports reliable extraction of year and month information for later analysis.

## 2.5 Creating Derived Features

After correcting the date datatype, I created a small number of derived variables that are useful for later analysis.

`Age Group` converts individual ages into broader customer segments, making age-based comparisons easier to interpret. `Purchase Year` and `Purchase Month` are extracted from `Purchase Date` so that the transaction data can also be examined from a time-based perspective.

These features are created from existing information and do not replace the original columns.

```python
df["Age Group"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)

df["Purchase Year"] = df["Purchase Date"].dt.year
df["Purchase Month"] = df["Purchase Date"].dt.month_name()

display(
    df[
        ["Age", "Age Group", "Purchase Date", "Purchase Year", "Purchase Month"]
    ].head()
)
```

### Result

Three derived variables were added successfully:

- `Age Group` — groups customers into five age segments.
- `Purchase Year` — represents the year of each transaction.
- `Purchase Month` — represents the month name of each transaction.

The dataset therefore increases from **26 original columns to 29 columns**, while the number of transaction records remains unchanged at 10,000.

## 2.6 Basic Filtering and Aggregation

To demonstrate how the processed data can be selected and summarized, I performed a simple filtering operation and a category-level aggregation.

The filtering step creates a separate DataFrame containing online transactions without modifying the original dataset. The aggregation then calculates the average Purchase Amount for each product category, providing an initial category-level spending comparison.

```python
online_df = df[df["Online/Offline"] == "Online"]

category_spending = (
    df.groupby("Category")["Purchase Amount (₹)"]
      .mean()
      .sort_values(ascending=False)
)

print("Number of Online Transactions:", len(online_df))

print("\n========== AVERAGE PURCHASE AMOUNT BY CATEGORY ==========")
print(category_spending.round(2))
```

### Observation

The filtered data contains **7,751 online transactions**.

The initial category-level aggregation shows that average Purchase Amount is highest for **Footwear (₹2,784.51)**, followed by **Clothing (₹1,671.25)** and **Accessories (₹1,220.53)**.

This is only an initial aggregation. Category-level spending is examined more carefully in Phase 3 using transaction counts, mean, and median values before drawing business conclusions.

## Analysis 6 — Discount vs Purchase Amount

**Analysis Type:** Bivariate Analysis

### Research Question
Is there a linear relationship between Discount Percentage and Purchase Amount, and how strong is that relationship?

### Purpose
The purpose of this analysis is to examine whether transactions with higher discount percentages are associated with higher or lower Purchase Amounts.

Discounting is an important part of retail pricing and promotional activity. However, observing that discounted transactions have different spending values does not automatically mean that discounts caused the difference. Therefore, this analysis focuses specifically on measuring the direction and strength of the statistical association between Discount Percentage and Purchase Amount.

### Variables Used
`Discount (%)` — a numerical variable representing the discount percentage applied to each transaction.

`Purchase Amount (₹)` — a numerical variable representing the monetary value of each transaction.

### Method Selected
Pearson correlation analysis using the Pandas `corr()` method.

The Pearson correlation coefficient measures the strength and direction of the linear relationship between two numerical variables. Its value ranges from -1 to +1.

A positive value indicates that the variables tend to move in the same direction, while a negative value indicates that they tend to move in opposite directions. Values closer to zero indicate a weaker linear relationship.

### Why Is This Method Used?
Both Discount Percentage and Purchase Amount are numerical variables, and the research question focuses on whether a linear association exists between them.

Pearson correlation provides a single coefficient that makes it possible to describe both the direction and approximate strength of this relationship.

This makes it useful as an initial statistical measure before examining the relationship visually or dividing discounts into broader levels.

### Alternative Methods
Spearman rank correlation could be used to measure a monotonic relationship without relying specifically on a linear relationship.

A scatter plot could also be used to visually examine how Purchase Amount changes across different discount percentages and whether the observations follow a clear pattern.

Discount Percentage could additionally be divided into groups and the average or median Purchase Amount compared across those groups.

### Why Was Pearson Correlation Selected?
Pearson correlation is suitable for the initial analysis because both variables are numerical and the immediate objective is to measure their linear association.

Spearman correlation would be useful if the analysis specifically focused on ranked or monotonic relationships. A scatter plot is valuable for visual inspection but does not provide the same concise numerical measure of linear association.

A grouped discount analysis is performed later to investigate whether the spending pattern differs across broader discount levels.

### What Should We Observe?
The main value to observe is the Pearson correlation coefficient.

We should examine whether the coefficient is positive or negative and whether its magnitude indicates a weak, moderate, or strong linear relationship.

The result should be interpreted as an association only. Correlation by itself does not establish that changing the discount percentage causes a change in Purchase Amount.

```python
discount_correlation = df[
    ["Discount (%)", "Purchase Amount (₹)"]
].corr().iloc[0, 1]

print("\n========== DISCOUNT VS PURCHASE AMOUNT ==========")
print("Pearson Correlation:", round(discount_correlation, 3))
```

### Observation

The Pearson correlation coefficient between `Discount (%)` and `Purchase Amount (₹)` is approximately **0.165**.

The coefficient is positive, indicating that higher discount percentages are associated with slightly higher Purchase Amounts in the dataset. However, the value is relatively close to zero.

### Interpretation

A correlation of approximately 0.165 represents a **weak positive linear relationship** between Discount Percentage and Purchase Amount.

This means that Purchase Amount shows some tendency to increase as Discount Percentage increases, but the relationship is not strong. A large amount of variation in Purchase Amount therefore remains unexplained by Discount Percentage alone.

The result should not be interpreted as evidence that increasing a discount directly causes customers to spend more. Product category, quantity, original pricing, customer behaviour, and other transaction characteristics may also contribute to Purchase Amount.

### Key Insight

Discount Percentage has only a **weak positive linear association** with Purchase Amount (`r ≈ 0.165`).

Therefore, discount level alone does not strongly differentiate transaction value within this dataset.

### Business Relevance

The result suggests that retail discount decisions should not be based only on the assumption that larger discounts will automatically produce proportionally higher transaction values.

Although higher discounts are associated with slightly higher Purchase Amounts, the relationship is weak. A more complete evaluation would require factors such as quantity sold, product category, cost, profit margin, and incremental sales.

Since cost and profit information are not available in the dataset, this analysis cannot determine whether higher discounts are financially beneficial.

### Limitation / Caution

Pearson correlation measures only linear association and can be affected by the distribution of the variables and unusual observations.

More importantly, correlation does not establish causation. The observed coefficient describes how the two variables are associated within this dataset; it does not prove that discounts caused the observed Purchase Amounts.

- A negative coefficient indicates that they tend to move in opposite directions.

- Larger absolute values indicate stronger linear relationships.

### Why Is This Method Used?

## Analysis 13 — Discount Efficiency & Customer Spending Response

**Analysis Type:** Segmented Bivariate Business Analysis

### Research Question
How does Purchase Amount vary across different Discount Levels, and does customer spending continue to increase at the same rate as discount percentage becomes higher?

### Purpose
The purpose of this analysis is to extend the earlier correlation analysis between Discount Percentage and Purchase Amount.

Analysis 6 identified a weak positive Pearson correlation between Discount Percentage and Purchase Amount. However, a single correlation coefficient summarizes the overall linear relationship and does not show whether the spending pattern changes across different parts of the discount range.

To investigate this further, Discount Percentage is divided into broader levels and Purchase Amount is compared across these groups. This helps determine whether higher discount levels are associated with progressively higher transaction values or whether the additional difference becomes smaller at the highest discount levels.

### Variables Used
`Discount (%)` — a numerical variable representing the discount percentage applied to each transaction.

`Purchase Amount (₹)` — a numerical variable representing the monetary value of each transaction.

A temporary analytical variable called `Discount Level` is created by grouping Discount Percentage into four ranges:

- Low: 0–10%
- Moderate: 11–30%
- High: 31–50%
- Very High: 51–70%

### Method Selected
Discount segmentation using Pandas `pd.cut()`, followed by grouped descriptive analysis using `groupby()` and `agg()`.

For each Discount Level, the following measures are calculated:

- `count` — number of transactions
- `mean` — average Purchase Amount
- `median` — middle Purchase Amount

### Why Is This Method Used?
Grouping the continuous Discount Percentage variable into broader ranges makes it possible to compare customer spending at different levels of discount intensity.

This approach provides information that is not visible from the Pearson correlation coefficient alone. In particular, it allows us to examine whether the difference in Purchase Amount remains similar as discounts increase or whether the pattern begins to flatten at higher discount levels.

Both mean and median are included because Purchase Amount has a right-skewed distribution.

### Alternative Methods
Discount Percentage could be analyzed directly using Pearson or Spearman correlation.

A scatter plot could also be used to examine the relationship visually.

Another approach would be to calculate average Purchase Amount for every individual discount percentage instead of creating broader groups.

### Why Was Discount Segmentation Selected?
Pearson correlation has already been used to measure the overall linear association, while a scatter plot is used later for visual inspection.

Segmenting discounts provides an additional business-oriented perspective because it allows transaction values to be compared across understandable discount ranges.

Using every individual discount percentage would create a more detailed but less concise output. Broader groups make the general spending pattern easier to compare and communicate.

### What Should We Observe?
We should compare the transaction count, mean Purchase Amount, and median Purchase Amount across the four Discount Levels.

Particular attention should be given to the change in average Purchase Amount from one discount level to the next.

If the increase becomes much smaller at the highest discount level, it may indicate that the positive spending pattern is flattening rather than continuing at the same rate.

```python
df["Discount Level"] = pd.cut(
    df["Discount (%)"],
    bins=[-1, 10, 30, 50, 70],
    labels=["Low", "Moderate", "High", "Very High"]
)

discount_analysis = (
    df.groupby("Discount Level", observed=True)["Purchase Amount (₹)"]
      .agg(["count", "mean", "median"])
)

discount_analysis["mean_difference"] = (
    discount_analysis["mean"].diff()
)

print("\n========== DISCOUNT LEVEL VS PURCHASE AMOUNT ==========")
print(discount_analysis.round(2))
```

### Observation

Purchase Amount differs across the four Discount Levels:

- **Low (0–10%):** 2,213 transactions, mean ₹1,305.31, median ₹845.00
- **Moderate (11–30%):** 2,521 transactions, mean ₹1,811.71, median ₹1,181.00
- **High (31–50%):** 2,600 transactions, mean ₹2,146.11, median ₹1,468.50
- **Very High (51–70%):** 2,666 transactions, mean ₹2,199.17, median ₹1,448.50

The average Purchase Amount increases across the four discount levels. However, the size of the increase becomes progressively smaller:

- Low → Moderate: approximately **+₹506.40**
- Moderate → High: approximately **+₹334.40**
- High → Very High: approximately **+₹53.06**

The median also increases from Low to High, but decreases slightly from ₹1,468.50 in the High group to ₹1,448.50 in the Very High group.

### Interpretation

The grouped analysis provides additional context to the weak positive correlation identified earlier.

Transactions in higher discount groups generally have higher average Purchase Amounts. However, the increase is not equally strong across the full discount range.

The average Purchase Amount increases substantially from Low to Moderate and from Moderate to High discounts, but the additional difference between High and Very High discounts is only about ₹53.06.

The median shows an even clearer sign of flattening at the highest discount level because it decreases slightly from High to Very High.

Therefore, the relationship between discount level and transaction value does not appear to increase proportionally across the entire discount range.

### Key Insight

**Higher discount levels are associated with higher average transaction values, but the additional spending difference becomes very small between the High and Very High discount groups.**

This suggests a flattening pattern at the highest discount range rather than a continuously increasing spending response of the same magnitude.

### Business Relevance

The result indicates that very high discounts should not automatically be assumed to produce proportionally larger transaction values.

From a retail perspective, the relatively small difference between the High and Very High groups makes the highest discount range an important area for further evaluation.

A business could compare these discount levels using additional measures such as units sold, product category, original price, cost, gross margin, and incremental sales before deciding whether very high discounts provide sufficient financial value.

### Limitation / Caution

The Discount Level boundaries used in this analysis are analyst-defined ranges created for comparison. They are not naturally occurring customer groups or experimental treatment groups.

The analysis is observational and does not establish that higher discounts caused higher Purchase Amounts. Other variables such as product category, quantity, pricing, or customer purchasing behaviour may contribute to the observed pattern.

In addition, the dataset does not contain product cost or profit-margin information. Therefore, this analysis evaluates transaction-value response only and cannot determine discount profitability or the financially optimal discount level.



## Analysis 14 — Monthly Purchase Trend

**Analysis Type:** Time-Based Analysis

### Research Question
How do transaction volume and average Purchase Amount vary across calendar months?

### Why This Analysis Is Added
`Purchase Month` was created during feature engineering. Using it in the analysis demonstrates that the derived time feature has a practical analytical purpose and helps identify possible monthly or seasonal purchasing patterns.

### Method
Transactions are grouped by month to calculate transaction count, total Purchase Amount, and average Purchase Amount. Months are ordered chronologically rather than alphabetically.


```python
month_order = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

monthly_analysis = (
    df.groupby("Purchase Month")["Purchase Amount (₹)"]
      .agg(Transaction_Count="count", Total_Purchase_Amount="sum", Average_Purchase_Amount="mean")
      .reindex(month_order)
      .dropna(how="all")
)

print("========== MONTHLY PURCHASE TREND ==========")
display(monthly_analysis.round(2))

print("Peak month by transaction count:", monthly_analysis["Transaction_Count"].idxmax())
print("Peak month by total purchase amount:", monthly_analysis["Total_Purchase_Amount"].idxmax())
print("Highest month by average purchase amount:", monthly_analysis["Average_Purchase_Amount"].idxmax())
```

Use this output to distinguish **volume**, **total value**, and **average value**. A month with the most transactions may not be the month with the highest average Purchase Amount. This distinction strengthens the business interpretation of time-based demand.

### 4.12 Discount Level vs Customer Spending Response

**Visualization Type:** Ordered Bar Chart

#### Research Question
How does average Purchase Amount vary across increasing discount levels, and does the spending increase remain substantial at progressively higher discount bands?

#### Purpose
Visualization 4.7 showed a weak transaction-level relationship between Discount Percentage and Purchase Amount.

Phase 3 then grouped discounts into broader levels and found that average Purchase Amount increased across the groups, but the additional increase became progressively smaller.

This visualization presents that grouped spending pattern in a more business-interpretable form.

#### Variables Used
- `Discount Group` — derived categorical variable representing discount ranges.
- `Purchase Amount (₹)`

The groups are:

- Low (0–10%)
- Moderate (11–30%)
- High (31–50%)
- Very High (51–70%)

#### Graph Selected
Ordered Bar Chart using Matplotlib.

#### Why Is a Bar Chart Used?
The Discount Groups are discrete categories with a meaningful low-to-high order. A bar chart makes their average Purchase Amounts easy to compare.

#### Alternative Graph
The scatter plot in Visualization 4.7 shows individual Discount Percentage and Purchase Amount observations.

#### Why Was the Ordered Bar Chart Selected?
The scatter plot shows transaction-level variability, while this graph answers a different question by comparing average spending across broader discount ranges.

#### What Should We Observe?
We should examine whether average Purchase Amount increases across the groups and whether the size of the increase remains similar as discounts become larger.

```python
discount_order = [
    "Low (0-10%)",
    "Moderate (11-30%)",
    "High (31-50%)",
    "Very High (51-70%)"
]

df["Discount Group"] = pd.cut(
    df["Discount (%)"],
    bins=[-1, 10, 30, 50, 70],
    labels=discount_order
)

discount_group_avg = (
    df.groupby("Discount Group", observed=True)["Purchase Amount (₹)"]
      .mean()
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    discount_group_avg.index.astype(str),
    discount_group_avg.values
)

plt.title("Average Purchase Amount by Discount Level")
plt.xlabel("Discount Group")
plt.ylabel("Average Purchase Amount (₹)")

plt.bar_label(
    bars,
    fmt="₹%.2f",
    padding=3
)

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

#### Observation

The bar chart shows that average Purchase Amount increases across the four Discount Groups:

Low (0–10%): ₹1,305.31  
Moderate (11–30%): ₹1,811.71  
High (31–50%): ₹2,146.11  
Very High (51–70%): ₹2,199.17

The increase between consecutive groups becomes progressively smaller:

Low → Moderate: +₹506.40  
Moderate → High: +₹334.40  
High → Very High: +₹53.06

#### Interpretation

The visualization shows a clear increase in average transaction value from the Low to Moderate and Moderate to High discount groups.

However, the increase becomes much smaller between the High and Very High groups. Average Purchase Amount rises by only ₹53.06 at the highest transition, producing a visible flattening in the bar heights.

This complements the Phase 3 analysis. The overall relationship between Discount Percentage and Purchase Amount was weak (`r ≈ 0.165`), while the grouped analysis shows that average spending differs across broader discount ranges. These findings are not contradictory because one measures the transaction-level linear relationship and the other compares group averages.

#### Key Insight

Higher Discount Groups are associated with higher average Purchase Amounts, but the additional increase becomes progressively smaller. In particular, the difference between the High and Very High groups is only ₹53.06.

#### Business Relevance

The relatively small difference between the High and Very High groups suggests that these two discount ranges deserve closer comparison using additional measures such as transaction volume, Quantity, Product Category, product cost, and profit margin.

Very High discounts should not automatically be assumed to correspond to proportionally higher transaction values.

#### Caution

The Discount Groups are analyst-defined categories, and the analysis describes associations rather than cause-and-effect relationships.

The dataset does not contain product cost or profit-margin information. Therefore, this analysis cannot determine whether a particular discount range is more profitable or identify an optimal discount level.

## Limitations and Future Scope

### Limitations

Although the project identified several meaningful patterns in customer shopping behaviour, the findings should be interpreted within the limitations of the available data and analytical methods.

- **Profitability cannot be measured directly:** The dataset contains Purchase Amount and Discount Percentage but does not include product cost, profit margin, or other financial measures. Therefore, the analysis cannot determine which Product Category, Shopping Channel, or Discount Group is most profitable.

- **The analysis identifies associations rather than causation:** Relationships such as Discount Percentage with Purchase Amount and Delivery Time with Review Rating are observational. They do not establish that one variable directly causes changes in another.

- **Customer-level analysis remains descriptive:** The enhanced project includes purchase-frequency and previous-purchase profiling, but advanced customer-level measures such as full RFM scores, CLV, cohort retention, and churn are not calculated.

- **Some group comparisons have unequal sample sizes:** For example, Review Rating 1 contains only 56 transactions, making its spending statistics less stable than those of larger rating groups.

- **Correlation analysis is limited to linear relationships:** Pearson correlation measures linear association. Important non-linear relationships may therefore not be fully represented by the correlation coefficients.

- **Analytical groups were defined for interpretation:** Age Groups and Discount Groups were created during the analysis to make comparisons easier. Different grouping boundaries could produce somewhat different aggregated results.

- **Statistical outliers were retained:** Several high-value Purchase Amounts exceeded the standard IQR boundary. These observations were retained because there was no evidence that they were data errors, but they can influence averages and other statistical measures.

### Future Scope

The current analysis can be extended in several directions if additional analysis or data becomes available.

- **Customer-level segmentation:** Customer IDs and purchase history could be used to develop customer-level measures such as Recency, Frequency, and Monetary value (RFM) and identify different customer segments.

- **Advanced time-series analysis:** The enhanced project includes monthly trend analysis. Future work could extend this into year-over-year comparisons, forecasting, seasonality decomposition, and campaign-period analysis where the date coverage is sufficient.

- **Basket and product-combination analysis:** Since Quantity showed the clearest positive association with Purchase Amount, future analysis could examine frequently purchased product combinations, basket composition, bundling, and cross-selling opportunities.

- **Statistical significance testing:** Group differences identified during EDA could be evaluated using appropriate statistical tests to determine whether observed differences are statistically significant rather than relying only on descriptive comparisons.

- **Multivariable modelling:** Regression or other analytical models could be used to examine Purchase Amount or Review Rating while considering several variables simultaneously.

- **Profitability and discount effectiveness:** Adding product cost, margin, and promotional cost data would allow discount levels and Product Categories to be evaluated using profitability and return-on-promotion measures rather than transaction value alone.

- **Delivery and return analysis:** Delivery Time, Delivery Speed, Shipping Charge, Return Status, Product Category, and Review Rating could be studied together to develop a deeper understanding of operational performance and customer experience.

Overall, these extensions could build on the current exploratory analysis and provide a more detailed customer-level, operational, and financial understanding of shopping behaviour.
